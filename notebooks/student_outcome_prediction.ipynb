# Student Academic Outcome Prediction

Predict the supplied dataset's `Target` (`Dropout`, `Enrolled`, `Graduate`) using only information available at enrollment. All computation uses the assignment-provided `data/data.csv`; no alternative dataset is loaded.

## Leakage prevention
Remove only the 12 listed first- and second-semester performance variables. They are measured after enrollment and unavailable at the intended prediction time.

## EDA
Inspect target distribution and enrollment-time age / tuition-status patterns. Treat these as descriptive associations, not causal claims.

## Split and preprocessing
Use a stratified 80/20 split and fixed seed. One-hot encode coded categories; impute missing values and scale numeric measurements inside the scikit-learn Pipeline. Learned preprocessing is fitted only on training folds / training data.

## Models and selection
Compare balanced Logistic Regression as a linear baseline with a balanced Random Forest ensemble. Select using stratified cross-validation macro F1 on training data only. Report both on the held-out test set without using test results to retune or select.

## Interpretation and limitations
Feature importance indicates model-specific predictive association, not causation. The dataset has uneven target frequencies and results may not generalize beyond this supplied dataset. Verify original dataset attribution/license before redistribution.

See `README.md` for executed metrics, per-class findings, the sample record, setup, and further limitations.

```python
# Imports for EDA, preprocessing, model selection, and evaluation
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.compose import ColumnTransformer
from sklearn.ensemble import RandomForestClassifier
from sklearn.impute import SimpleImputer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix,
    f1_score,
    precision_score,
    recall_score,
)
from sklearn.model_selection import (
    StratifiedKFold,
    cross_val_score,
    train_test_split,
)
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler

RANDOM_STATE = 42
TARGET_COLUMN = "Target"
CLASS_LABELS = ["Dropout", "Enrolled", "Graduate"]

# Stage 3: exclude exactly the supplied post-enrollment variables.
semester_performance_columns = [
    "Curricular units 1st sem (credited)",
    "Curricular units 1st sem (enrolled)",
    "Curricular units 1st sem (evaluations)",
    "Curricular units 1st sem (approved)",
    "Curricular units 1st sem (grade)",
    "Curricular units 1st sem (without evaluations)",
    "Curricular units 2nd sem (credited)",
    "Curricular units 2nd sem (enrolled)",
    "Curricular units 2nd sem (evaluations)",
    "Curricular units 2nd sem (approved)",
    "Curricular units 2nd sem (grade)",
    "Curricular units 2nd sem (without evaluations)",
]
missing_semester_columns = [c for c in semester_performance_columns if c not in df.columns]
assert not missing_semester_columns, f"Expected leakage columns not found: {missing_semester_columns}"
assert TARGET_COLUMN in df.columns
assert set(df[TARGET_COLUMN].dropna().unique()) == set(CLASS_LABELS)

primary_feature_columns = [
    c for c in df.columns
    if c != TARGET_COLUMN and c not in semester_performance_columns
]
print("Excluded post-enrollment variables:")
print(*semester_performance_columns, sep="\n")
print(f"\nExcluded: {len(semester_performance_columns)}; retained enrollment-time predictors: {len(primary_feature_columns)}")
```

```python
# Stage 4: descriptive EDA; df remains unchanged.
fig, axes = plt.subplots(1, 3, figsize=(16, 4))
target_counts = df[TARGET_COLUMN].value_counts().reindex(CLASS_LABELS)
target_counts.plot(kind="bar", ax=axes[0], color=["#d95f59", "#e6b655", "#4b9b72"])
axes[0].set_title("Target counts")
axes[0].set_xlabel("Final outcome")
axes[0].set_ylabel("Students")
axes[0].tick_params(axis="x", rotation=25)

for label in CLASS_LABELS:
    df.loc[df[TARGET_COLUMN] == label, "Age at enrollment"].plot(
        kind="hist", bins=20, alpha=0.45, ax=axes[1], label=label
    )
axes[1].set_title("Age at enrollment by outcome")
axes[1].set_xlabel("Age at enrollment")
axes[1].legend()

tuition_rates = pd.crosstab(
    df["Tuition fees up to date"], df[TARGET_COLUMN], normalize="index"
).reindex(columns=CLASS_LABELS)
tuition_rates.plot(kind="bar", stacked=True, ax=axes[2], colormap="Set2")
axes[2].set_title("Outcome shares by tuition status")
axes[2].set_xlabel("Tuition fees up to date code")
axes[2].set_ylabel("Share within status")
axes[2].tick_params(axis="x", rotation=0)
plt.tight_layout()
plt.show()
print("Target percentages:")
display((df[TARGET_COLUMN].value_counts(normalize=True).reindex(CLASS_LABELS) * 100).rename("percent"))
print("Descriptive outcomes by tuition-status code (row percentages):")
display(tuition_rates)
```

```python
# Stages 5–7: split, preprocessing, and model pipelines.
X = df.loc[:, primary_feature_columns].copy()
y = df[TARGET_COLUMN].copy()
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=RANDOM_STATE, stratify=y
)
print(f"Training rows: {len(X_train)}; test rows: {len(X_test)}")
print("Training class shares:")
display(y_train.value_counts(normalize=True).reindex(CLASS_LABELS).rename("share"))
print("Test class shares:")
display(y_test.value_counts(normalize=True).reindex(CLASS_LABELS).rename("share"))

# Stage 6: all integer-coded categories are treated as nominal codes (not continuous magnitudes).
categorical_features = [
    "Marital status", "Application mode", "Application order", "Course",
    "Daytime/evening attendance", "Previous qualification", "Nacionality",
    "Mother's qualification", "Father's qualification", "Mother's occupation",
    "Father's occupation", "Displaced", "Educational special needs", "Debtor",
    "Tuition fees up to date", "Gender", "Scholarship holder", "International",
]
numerical_features = [c for c in primary_feature_columns if c not in categorical_features]
assert set(categorical_features + numerical_features) == set(primary_feature_columns)
print("Categorical / encoded-category predictors:", categorical_features)
print("Numerical measurement predictors:", numerical_features)

numeric_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler()),
])
categorical_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("onehot", OneHotEncoder(handle_unknown="ignore")),
])
preprocessor = ColumnTransformer([
    ("numeric", numeric_pipeline, numerical_features),
    ("categorical", categorical_pipeline, categorical_features),
])

# Stage 7: a scaled linear baseline and a nonlinear tree ensemble.
model_pipelines = {
    "Logistic Regression": Pipeline([
        ("preprocess", preprocessor),
        ("model", LogisticRegression(
            max_iter=2000, class_weight="balanced", random_state=RANDOM_STATE
        )),
    ]),
    "Random Forest": Pipeline([
        ("preprocess", preprocessor),
        ("model", RandomForestClassifier(
            n_estimators=300, class_weight="balanced", random_state=RANDOM_STATE,
            n_jobs=-1, min_samples_leaf=2
        )),
    ]),
}
```

```python
# Choose by stratified CV macro F1 on training data only; the test split is untouched here.
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=RANDOM_STATE)
cv_scores = {}
for model_name, pipeline in model_pipelines.items():
    scores = cross_val_score(
        pipeline, X_train, y_train, cv=cv, scoring="f1_macro", n_jobs=1
    )
    cv_scores[model_name] = scores
cv_summary = pd.DataFrame({
    "CV macro F1 mean": {name: values.mean() for name, values in cv_scores.items()},
    "CV macro F1 std": {name: values.std() for name, values in cv_scores.items()},
}).sort_values("CV macro F1 mean", ascending=False)
print("Training-only cross-validation results (selection metric: macro F1):")
display(cv_summary)
selected_model_name = cv_summary.index[0]
print(f"Selected using training-only CV macro F1: {selected_model_name}")
```

```python
# Stages 8–10: fit on training split and evaluate once on the held-out test set.
fitted_models = {}
test_predictions = {}
comparison_rows = []
per_class_results = {}
confusion_matrices = {}
for model_name, pipeline in model_pipelines.items():
    pipeline.fit(X_train, y_train)
    predictions = pipeline.predict(X_test)
    fitted_models[model_name] = pipeline
    test_predictions[model_name] = predictions
    report = classification_report(
        y_test, predictions, labels=CLASS_LABELS,
        output_dict=True, zero_division=0
    )
    per_class_results[model_name] = pd.DataFrame(
        {label: report[label] for label in CLASS_LABELS}
    ).T[["precision", "recall", "f1-score", "support"]]
    confusion_matrices[model_name] = confusion_matrix(
        y_test, predictions, labels=CLASS_LABELS
    )
    comparison_rows.append({
        "Model": model_name,
        "Accuracy": accuracy_score(y_test, predictions),
        "Precision (macro)": precision_score(
            y_test, predictions, labels=CLASS_LABELS, average="macro", zero_division=0
        ),
        "Recall (macro)": recall_score(
            y_test, predictions, labels=CLASS_LABELS, average="macro", zero_division=0
        ),
        "F1 (macro)": f1_score(
            y_test, predictions, labels=CLASS_LABELS, average="macro", zero_division=0
        ),
        "F1 (weighted)": f1_score(y_test, predictions, average="weighted", zero_division=0),
    })
comparison_table = pd.DataFrame(comparison_rows).set_index("Model")
print("Held-out test metrics (reporting only; not used for model selection):")
display(comparison_table)
for model_name in model_pipelines:
    print(f"\n{model_name} — per-class test metrics")
    display(per_class_results[model_name])
    print("Confusion matrix (rows=true, columns=predicted; class order:", CLASS_LABELS, ")")
    display(pd.DataFrame(
        confusion_matrices[model_name], index=CLASS_LABELS, columns=CLASS_LABELS
    ))

# Stage 10: the final model is the training-CV macro-F1 winner, not the test-set winner.
final_model = fitted_models[selected_model_name]
print("Final model selected from training-only CV:", selected_model_name)
print("The held-out test comparison is reported as an unbiased assessment, not a selection step.")
```

```python
# Stage 11: tree impurity importance or mean absolute multiclass coefficient, as appropriate.
fitted_preprocessor = final_model.named_steps["preprocess"]
encoded_feature_names = fitted_preprocessor.get_feature_names_out()
fitted_estimator = final_model.named_steps["model"]
if hasattr(fitted_estimator, "feature_importances_"):
    importances = fitted_estimator.feature_importances_
    importance_method = "Random Forest impurity-based feature importance"
elif hasattr(fitted_estimator, "coef_"):
    importances = np.abs(fitted_estimator.coef_).mean(axis=0)
    importance_method = "Mean absolute Logistic Regression coefficient across classes"
else:
    raise TypeError(f"No supported feature-importance method for {selected_model_name}")
feature_importance = pd.Series(
    importances, index=encoded_feature_names, name="importance"
).sort_values(ascending=False)
print(importance_method)
display(feature_importance.head(15).to_frame())
feature_importance.head(15).sort_values().plot(kind="barh", figsize=(9, 6))
plt.title(f"Top encoded predictors — {selected_model_name}")
plt.xlabel("Importance (method-specific; not causal)")
plt.tight_layout()
plt.show()
print("Feature importance indicates predictive association in this fitted model; it does not establish causation.")
```

```python
# Stage 12: held-out sample predictions through the same fitted pipeline.
sample_input = X_test.head(5).copy()
sample_output = pd.DataFrame({
    "predicted_class": final_model.predict(sample_input),
    "actual_class_for_verification": y_test.loc[sample_input.index].to_numpy(),
}, index=sample_input.index)
print("Held-out student input records:")
display(sample_input)
print("Predictions (actual labels shown only to verify these held-out examples):")
display(sample_output)

# Conclusion: consider class-level results and limitations before any operational use.
```

```python
# Verify the fitted pipeline's raw predictor columns contain no semester-performance leakage.
final_predictor_columns = list(final_model.feature_names_in_)
leakage_in_final_model = sorted(
    set(final_predictor_columns).intersection(semester_performance_columns)
)
assert not leakage_in_final_model, (
    f"Leakage columns found in final model predictors: {leakage_in_final_model}"
)
assert set(final_predictor_columns) == set(primary_feature_columns)
print("Leakage check passed: no semester-performance predictors are used.")

# Save and reload the complete fitted preprocessing + model pipeline.
from pathlib import Path
import joblib

model_path = Path("../models/student_academic_outcome_model.pkl")
model_path.parent.mkdir(parents=True, exist_ok=True)
joblib.dump(final_model, model_path)
reloaded_model = joblib.load(model_path)
reloaded_sample_predictions = reloaded_model.predict(sample_input)
np.testing.assert_array_equal(
    reloaded_sample_predictions,
    final_model.predict(sample_input),
)
print(f"Model save/reload verification passed: {model_path}")
display(pd.DataFrame({
    "reloaded_prediction": reloaded_sample_predictions,
    "actual_target": y_test.loc[sample_input.index].to_numpy(),
}, index=sample_input.index))
```
